```python
# import libraries
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path
```

```python
# load data
df1 = pd.read_csv("../data/climate-daily-st_johns.csv")
print(f"data 1 shape: {df1.shape}")
df2 = pd.read_csv("../data/climate-daily-mount_pearl.csv")
print(f"data 2 shape: {df2.shape}")

df_concat = pd.concat([df1, df2], ignore_index=True)
print(f"\nMerged dataset shape: {df_concat.shape}")
df_concat.head()
```

```python
df_concat['LOCAL_DATE_dt'] = pd.to_datetime(
    df_concat['LOCAL_DATE'], errors='coerce')
df_concat[["LOCAL_DATE_dt", "LOCAL_DATE"]]
```

```python
features = ['x', 'y', 'STATION_NAME', 'CLIMATE_IDENTIFIER', 'ID', 'LOCAL_DATE',
            'LOCAL_DATE_dt', 'PROVINCE_CODE', 'LOCAL_YEAR', 'LOCAL_MONTH', 'LOCAL_DAY',
            'MEAN_TEMPERATURE', 'MIN_TEMPERATURE', 'MAX_TEMPERATURE',
            'TOTAL_PRECIPITATION', 'TOTAL_RAIN', 'TOTAL_SNOW', 'SNOW_ON_GROUND',
            'COOLING_DEGREE_DAYS', 'HEATING_DEGREE_DAYS',]
```

```python
df_unflagged = df_concat[features]
df_unflagged.head()
```

```python
from sklearn.impute import KNNImputer

features_to_impute = ['MEAN_TEMPERATURE',
                      'MIN_TEMPERATURE',
                      'MAX_TEMPERATURE',
                      'TOTAL_PRECIPITATION',
                      'TOTAL_RAIN',
                      'TOTAL_SNOW',
                      'SNOW_ON_GROUND',
                      'COOLING_DEGREE_DAYS',
                      'HEATING_DEGREE_DAYS']

features_existing = [
    f for f in features_to_impute
    if f in df_unflagged.columns and pd.api.types.is_numeric_dtype(df_unflagged[f])
]

print(f"Imputing {len(features_existing)} numeric columns:")
print(features_existing)

# Extract only features
data = df_unflagged[features_existing].values

if data.dtype == 'object':
    raise ValueError("Data still contains non-numeric values!")

# Initialize and Impute data
imputer = KNNImputer(n_neighbors=5, weights='distance')
data_imputed = imputer.fit_transform(data)

# replace in main data
df_imputed = df_unflagged.copy()
df_imputed[features_existing] = pd.DataFrame(
    data_imputed,
    columns=features_existing,
    index=df_unflagged.index
)
```

```python
plt.figure(figsize=(12, 6))
sns.lineplot(data=df_imputed, x='LOCAL_YEAR', y='MEAN_TEMPERATURE')
plt.title('Temperature Over Time')
plt.xlabel('Date')
plt.ylabel('Temperature (°C)')

plt.xticks(rotation=45)
plt.tight_layout()
plt.savefig("../plots/mean_temp_over_time.png", dpi=300)
plt.show()
```
