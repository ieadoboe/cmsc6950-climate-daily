```python
# load data
df1 = pd.read_csv("../data/climate-daily-1.csv")
print(f"data 1 shape: {df1.shape}")
df2 = pd.read_csv("../data/climate-daily-2.csv")
print(f"data 1 shape: {df2.shape}")
df3 = pd.read_csv("../data/climate-daily-1840-2000-1.csv")
print(f"data 1 shape: {df3.shape}")
df4 = pd.read_csv("../data/climate-daily-1840-2000-2.csv")
print(f"data 1 shape: {df4.shape}")
# df_merge = pd.merge(df1, df2, on='y', how='left')
df_merge = pd.concat([df1, df2, df3, df4], ignore_index=True)
print(f"\nMerged dataset shape: {df_merge.shape}")
df_merge.head()
```

```python
df_merge['LOCAL_DATE_dt'] = pd.to_datetime(
    df_merge['LOCAL_DATE'], errors='coerce')
df_merge[["LOCAL_DATE_dt", "LOCAL_DATE"]]
```

```python
df_merge.columns
```

```python
features = ['x', 'y', 'STATION_NAME', 'CLIMATE_IDENTIFIER', 'ID', 'LOCAL_DATE',
            'LOCAL_DATE_dt', 'PROVINCE_CODE', 'LOCAL_YEAR', 'LOCAL_MONTH', 'LOCAL_DAY',
            'MEAN_TEMPERATURE', 'MIN_TEMPERATURE', 'MAX_TEMPERATURE',
            'TOTAL_PRECIPITATION', 'TOTAL_RAIN', 'TOTAL_SNOW', 'SNOW_ON_GROUND',
            'COOLING_DEGREE_DAYS', 'HEATING_DEGREE_DAYS',]
```

```python
df_unflagged = df_merge[features]
df_unflagged.head()
```

```python
df_unflagged[["LOCAL_DATE_dt"]].max()
```

```python
# Missing values
print(df_unflagged.isnull().sum())
```

```python
# visualize missing values
sns.heatmap(df_unflagged.isnull(), cbar=False)
```

```python
df_unflagged.shape
```

```python
from sklearn.impute import KNNImputer

features_to_impute = ['x', 'y', 'STATION_NAME', 'CLIMATE_IDENTIFIER', 'ID', 'LOCAL_DATE', 'LOCAL_DATE_dt', 'PROVINCE_CODE', 'LOCAL_YEAR', 'LOCAL_MONTH', 'LOCAL_DAY',
                      'MEAN_TEMPERATURE', 'MIN_TEMPERATURE', 'MAX_TEMPERATURE', 'TOTAL_PRECIPITATION', 'TOTAL_RAIN', 'TOTAL_SNOW', 'SNOW_ON_GROUND', 'COOLING_DEGREE_DAYS', 'HEATING_DEGREE_DAYS']

features_existing = [
    f for f in features_to_impute
    if f in df_unflagged.columns and pd.api.types.is_numeric_dtype(df_unflagged[f])
]

print(f"Imputing {len(features_existing)} numeric columns:")
print(features_existing)

# Extract only features
data = df_unflagged[features_existing].values

if data.dtype == 'object':
    raise ValueError("Data still contains non-numeric values!")

# Initialize and Impute data
imputer = KNNImputer(n_neighbors=5, weights='distance')
data_imputed = imputer.fit_transform(data)

# replace in main data
df_imputed = df_unflagged.copy()
df_imputed[features_existing] = pd.DataFrame(
    data_imputed,
    columns=features_existing,
    index=df_unflagged.index
)
```
